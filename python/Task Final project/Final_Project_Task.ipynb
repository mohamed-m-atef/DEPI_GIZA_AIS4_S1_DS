```python
def bank_account():
    """"
    This function creates a bank account with an initial balance and allows the user to deposit, withdraw, and check the balance.
    Returns:
        deposit: A function that allows the user to deposit money into the account.\n
        withdraw: A function that allows the user to withdraw money from the account.\n
        check_balance: A function that allows the user to check the current balance of the account.
    """
    print("Welcome to the bank account manager")
    print("Enter the balance of this account")
    balance = float(input())
    print(f"Account created with balance: {balance}")
    def deposit(amount):
        print("Enter the amount to deposit")
        amount = float(input())
        nonlocal balance
        balance = balance + amount
        print(f"New Balance is: {balance}")
        return balance
    def withdraw(amount):
        nonlocal balance
        print("Enter the amount to withdraw")
        amount = float(input())
        if amount > balance:
            return "Insufficient funds"
        else:
            balance = balance - amount
            print(f"New Balance is: {balance}")
            return balance
    def check_balance():
        return balance
    return deposit, withdraw, check_balance

deposit, withdraw, check_balance = bank_account()
deposit(1500)
withdraw(500)
print(f"Current Balance: {check_balance()}")

```

```python
def calc():
    """
    A simple calculator function that performs basic arithmetic operations based on user input.
    using float numbers and handles division by zero.
    """
    def add(a, b):
        return a + b
    def subtract(a, b):
        return a - b
    def multiply(a, b):
        return a * b
    def divide(a, b):
        if b != 0:
            return a / b
        else:
            return "Error! Division by zero."
    print("Welcome to the calculator!")
    print("Choose an operation:")
    print("1: Add")
    print("2: Subtract")
    print("3: Multiply")
    print("4: Divide")
    choice = input("Enter the number of the operation you want to perform 1: Add\n 2: Subtract\n 3: Multiply\n 4: Divide\n ")
    a = float(input("Enter the first number: "))
    b = float(input("Enter the second number: "))
    if choice == "1":
        c = add(a,b)
        print("the addition of",a,"and",b,"is equal",c)
    elif choice == "2":
        c = subtract(a,b)
        print("the subtraction of",a,"and",b,"is equal",c)
    elif choice == "3":
        c = multiply(a,b)
        print("the multiplication of",a,"and",b,"is equal",c)
    elif choice == "4":
        c = divide(a,b)
        print("the division of",a,"and",b,"is equal",c)
    else:
        print("Invalid choice! Please select a valid operation (1/2/3/4).")


calc()
```
