```python
import os
import random

def folder(folder_path, num_files):
    if not os.path.exists(folder_path):
        os.makedirs(folder_path)
        print(f"The directory '{folder_path}' has been created.")
    else:
        print(f"The directory '{folder_path}' already exists.")
    
    for i in range(num_files):
        file_path = os.path.join(folder_path, f"file_{i+1}.txt")
        with open(file_path, 'w') as f:
            f.write(f"This is file number {i+1}\n")
            f.write(f"Random number: {random.randint(1, 100)}\n")
folder("D:\\DEPI Data Scientist\\DS\\DEPI_GIZA_AIS4_S1_DS\\python\\Task assignment 2 (files)",5)

files = os.listdir("D:\\DEPI Data Scientist\\DS\\DEPI_GIZA_AIS4_S1_DS\\python\\Task assignment 2 (files)")
print("Files in directory:", files)
random.shuffle(files)
print("Shuffled files:", files)

half = len(files) // 2
to_delete = files[:half]
for file in to_delete:
    os.remove(os.path.join("D:\\DEPI Data Scientist\\DS\\DEPI_GIZA_AIS4_S1_DS\\python\\Task assignment 2 (files)", file))
    print(f"Deleted file: {file}")

remaining_files = os.listdir("D:\\DEPI Data Scientist\\DS\\DEPI_GIZA_AIS4_S1_DS\\python\\Task assignment 2 (files)")
print("Remaining files:", remaining_files)
```
